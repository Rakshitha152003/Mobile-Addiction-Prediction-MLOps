```python
# Outlier Analysis Using Boxplots
plt.figure(figsize=(14, 6))

sns.boxplot(data=df.select_dtypes(include="number"))

plt.title("Outlier Analysis of Numerical Features")
plt.xticks(rotation=90)
plt.show()
```

```python
# Outlier Detection Using IQR
numeric_df = df.select_dtypes(include="number").drop(columns=["Addiction_Level"])

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outlier_count = ((numeric_df < lower_limit) | (numeric_df > upper_limit)).sum()

print("Outliers in Each Numerical Feature:")
print(outlier_count)
```
